## Why repeated attempts need careful interpretation

HTB notes that wording and model versions affect outcomes. The same input can produce different responses, so one refusal or one success is only an observation about a particular setup and run. Do not treat repeated attempts as evidence that a method transfers to other models.

A token is a text piece processed by the model, such as a word fragment or punctuation. Longer prompts contain more tokens, but length itself does not create higher authority. Adversarial suffixes may look nonsensical because they are selected for model behavior rather than ordinary human meaning; they can be brittle and model-specific.

```python
# Record only redacted outcome labels from the in-scope HTB lab.
observations = []

def record_observation(strategy, channel, outcome, repeat_number=1):
    """Store a short outcome label without retaining prompt or secret text."""
    allowed = {
        "refused",
        "changed intended task",
        "disclosed (redacted)",
        "no visible change",
        "other (redacted)",
    }
    if outcome not in allowed:
        raise ValueError(f"Choose a redacted outcome from: {sorted(allowed)}")

    item = {
        "strategy": strategy,
        "channel": channel,  # for example: direct user message
        "outcome": outcome,
        "repeat_number": repeat_number,
    }
    observations.append(item)
    return item

# Example format only; it does not interact with the HTB service.
# record_observation("roleplay", "direct user message", "refused")
```
