## Two different goals

1. Preventing the model from ever following an injected instruction is difficult to guarantee.
2. Preventing a model mistake from exposing a secret or causing an unauthorized action is an application-design goal.

Prompt wording and filters can reduce risk, but the application should also limit the information and permissions available to the model and validate important actions outside the model. This distinction refines HTB's statement that avoiding LLMs is the only guaranteed way to avoid prompt injection itself.

```python
# Each row separates what a control can help with from its limitation.
mitigation_matrix = [
    {
        "control": "System prompt wording",
        "can_help_with": "guiding ordinary responses and stating the intended task",
        "does_not_guarantee": "resistance to direct or indirect prompt injection",
    },
    {
        "control": "Whitelist",
        "can_help_with": "limiting a narrow application to approved inputs",
        "does_not_guarantee": "a useful flexible assistant experience",
    },
    {
        "control": "Blacklist or similarity filter",
        "can_help_with": "blocking known words or close known patterns",
        "does_not_guarantee": "coverage of synonyms, obfuscation, or novel phrasing",
    },
    {
        "control": "Least privilege",
        "can_help_with": "limiting secrets and tools available to the model",
        "does_not_guarantee": "that the model never follows injected text",
    },
    {
        "control": "Human approval",
        "can_help_with": "catching risky recommendations before consequential actions",
        "does_not_guarantee": "a scalable substitute for technical controls",
    },
]

for row in mitigation_matrix:
    print(f"{row['control']}:")
    print(f"  Helps: {row['can_help_with']}")
    print(f"  Limit: {row['does_not_guarantee']}")
```

## Apply the controls to a system

For each use case, ask:

- Does the model need the secret or sensitive record to do its task?
- Can a model response directly change data, grant access, or trigger an action?
- Which permissions can be removed or narrowed?
- What trusted code verifies identity, policy, and output format?
- Which decisions require a human approval step?
- What evidence supports the claim that a control is actually enforced?

A claim in a system prompt is not evidence that the application enforces a permission.
