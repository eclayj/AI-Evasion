# Unsupervised learning: what does “similar” mean?

Companion to HTB Academy Module 290, Section 9. This notebook computes the section's Euclidean, Manhattan, cosine, and feature-scaling ideas with tiny invented vectors. It does **not** run a clustering or anomaly-detection algorithm from a later lesson.

Read $d_2(x,z)=\sqrt{\sum_j(x_j-z_j)^2}$ as *the square root of the sum over features j of x sub j minus z sub j, squared*. Read $d_1(x,z)=\sum_j|x_j-z_j|$ as *the sum of the absolute coordinate differences*.

```python
from math import sqrt

x, z = (1.0, 2.0), (4.0, 6.0)
coordinate_differences = [a - b for a, b in zip(x, z)]
# Formula: Euclidean = sqrt(sum_j (x_j-z_j)^2).
euclidean = sqrt(sum(delta ** 2 for delta in coordinate_differences))
# Formula: Manhattan = sum_j |x_j-z_j|.
manhattan = sum(abs(delta) for delta in coordinate_differences)
print('coordinate differences:', coordinate_differences)
print('Euclidean distance:', euclidean, 'Manhattan distance:', manhattan)
assert coordinate_differences == [-3.0, -4.0]
assert euclidean == 5.0 and manhattan == 7.0
```

## Cosine looks at direction, not straight-line gap

Read $\operatorname{cosim}(x,z)=(x\cdot z)/(\lVert x\rVert_2\lVert z\rVert_2)$ as *x dot z, divided by the length of x times the length of z*. A zero vector has no direction and makes this expression undefined.

```python
direction_a, direction_b = (1.0, 0.0), (1.0, 1.0)
dot_product = sum(a * b for a, b in zip(direction_a, direction_b))
length_a = sqrt(sum(value ** 2 for value in direction_a))
length_b = sqrt(sum(value ** 2 for value in direction_b))
assert length_a > 0 and length_b > 0  # Cosine is undefined for a zero vector.
cosine_similarity = dot_product / (length_a * length_b)
print(f'Cosine similarity: {cosine_similarity:.3f}')
assert abs(cosine_similarity - 1 / sqrt(2)) < 1e-12
```

## Units can dominate a distance

Imagine two observations with a first feature near 20–30 and a second near 50,000–60,000. The raw Euclidean distance mostly measures the second feature's units. Dividing the first column by 10 and the second by 10,000 is an **illustrative unit change**, not a fitted production scaler. A real scaler's settings must come from training data and be reused unchanged for new rows.

```python
row_a, row_b = (20.0, 50_000.0), (30.0, 60_000.0)
raw_distance = sqrt(sum((a - b) ** 2 for a, b in zip(row_a, row_b)))
units = (10.0, 10_000.0)  # Fixed, illustrative divisors for these two columns.
scaled_a = tuple(value / unit for value, unit in zip(row_a, units))
scaled_b = tuple(value / unit for value, unit in zip(row_b, units))
scaled_distance = sqrt(sum((a - b) ** 2 for a, b in zip(scaled_a, scaled_b)))
print(f'Raw distance: {raw_distance:.3f}; after unit scaling: {scaled_distance:.3f}')
assert raw_distance > 10_000
assert abs(scaled_distance - sqrt(2)) < 1e-12
```

A close pair under one representation need not be close under another. Unlabeled grouping is therefore an *analysis choice*, not discovery of guaranteed ground-truth classes. A rare point may be a harmless data error or legitimate event; an attack may resemble common traffic.
