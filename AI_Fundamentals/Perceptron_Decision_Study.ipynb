# Perceptron decisions: every term visible

Companion to HTB Academy Module 290, Section 17. This notebook reproduces the course's **given** tennis weights. It performs a forward calculation only; no learning or optimizer is implied.

Read z = Σᵢ wᵢxᵢ + b as **z equals the sum of w sub i times x sub i, plus bias b**. Read y = f(z) as **y equals f of z**. HTB's step function returns 1 only when z is strictly greater than zero.

```python
feature_names = ('outlook', 'temperature', 'humidity', 'wind')
weights = (0.3, 0.2, -0.4, -0.2)  # Given by HTB, not learned here.
bias = 0.1

def step_activation(score):
    """Return the HTB binary decision, including zero on the non-positive side."""
    return 1 if score > 0 else 0

def perceptron_decision(inputs):
    """Expose each w_i*x_i term, their sum plus bias, and the final step."""
    if len(inputs) != len(weights):
        raise ValueError('Expected exactly four encoded weather features')
    weighted_terms = [weight * value for weight, value in zip(weights, inputs)]
    score = sum(weighted_terms) + bias  # z = Σ_i w_i*x_i + b
    decision = step_activation(score)  # y = f(z)
    return weighted_terms, score, decision

assert step_activation(-0.1) == 0
assert step_activation(0.0) == 0  # The threshold is strictly greater than zero.
assert step_activation(0.1) == 1
```

## Work the course's tennis day

HTB encodes sunny=0, mild=1, high humidity=0, weak wind=0. The four products below are 0, 0.2, 0, 0; adding bias 0.1 gives z=0.3. Since 0.3>0, y=1 means play tennis.

```python
htb_day = (0, 1, 0, 0)
terms, score, decision = perceptron_decision(htb_day)
for name, value, weight, term in zip(feature_names, htb_day, weights, terms):
    print(f'{name}: {weight:+.1f} * {value} = {term:+.1f}')
print(f'weighted terms {sum(terms):+.1f} + bias {bias:+.1f} = score {score:+.1f}')
print('play tennis' if decision == 1 else 'do not play tennis')
assert terms == [0.0, 0.2, -0.0, -0.0]
assert abs(score - 0.3) < 1e-12 and decision == 1
```

## Change inputs, not the weights

Hot=0, high humidity=0, and strong wind=1 produce a negative score with the same fixed weights. This illustrates a boundary crossing, not the quality of the model. The numeric weather-category encoding assumes an order and spacing that need not be appropriate for real data.

```python
other_day = (0, 0, 1, 1)  # Sunny, hot, normal humidity, strong wind.
other_terms, other_score, other_decision = perceptron_decision(other_day)
print('other-day terms:', other_terms, 'score:', other_score, 'decision:', other_decision)
assert abs(other_score - (-0.5)) < 1e-12
assert other_decision == 0
```

## Why one perceptron cannot implement XOR

For binary inputs x₁ and x₂, XOR wants 0,1,1,0 on (0,0),(1,0),(0,1),(1,1). To make (0,0) output 0, bias b must be at most 0. The middle two required ones imply w₁+b>0 and w₂+b>0. Adding them forces w₁+w₂+b>−b≥0, which incorrectly makes (1,1) output 1. This is a proof about every possible choice of one perceptron's weights, not a failed parameter search. Multiple units with suitable nonlinearities are a later topic.
