# Reinforcement learning: immediate versus later rewards

Companion to HTB Academy Module 290, Section 13. HTB introduces agent, environment, state, action, reward, policy, value, discount factor, and episodic/continuing tasks. This small notebook only illustrates **discounting a finite reward trace**; it does not implement or train an RL algorithm.

Read γ as **gamma**, the factor applied once per step into the future. In the explanatory formula G = r₀ + γr₁ + γ²r₂, say **G equals r sub zero plus gamma times r sub one plus gamma squared times r sub two**. The subscript tells us *when* a reward arrives; the exponent tells us how strongly later reward is discounted.

```python
def discounted_total(rewards, gamma):
    """Add a finite reward trace, weighting step t by gamma**t."""
    if not 0 <= gamma <= 1:
        raise ValueError('This example uses gamma between zero and one')
    # Enumerate supplies t=0 for now, t=1 for one step later, and so on.
    contributions = [reward * gamma**t for t, reward in enumerate(rewards)]
    return sum(contributions), contributions

future_goal = [0, 10]  # No reward now; 10 one step later.
short_horizon, short_parts = discounted_total(future_goal, 0.2)
long_horizon, long_parts = discounted_total(future_goal, 0.9)
print('gamma=0.2:', short_parts, 'total:', short_horizon)
print('gamma=0.9:', long_parts, 'total:', long_horizon)
assert short_horizon == 2
assert long_horizon == 9
```

## Compare two actions in the same invented state

Action A gives 3 immediately and nothing later. Action B gives nothing immediately but reaches 10 one step later. The *policy* would have to choose one of these actions from the current state. We can compare finite returns, but we have not learned a value function: we supplied both possible reward sequences ourselves.

```python
immediate_action = [3, 0]
goal_action = [0, 10]
for gamma in (0.2, 0.9):
    immediate_value, _ = discounted_total(immediate_action, gamma)
    goal_value, _ = discounted_total(goal_action, gamma)
    preferred = 'immediate action' if immediate_value > goal_value else 'goal action'
    print(f'gamma={gamma}: immediate={immediate_value:g}, goal={goal_value:g}; prefer {preferred}')
assert discounted_total(immediate_action, 0.2)[0] > discounted_total(goal_action, 0.2)[0]
assert discounted_total(immediate_action, 0.9)[0] < discounted_total(goal_action, 0.9)[0]
```

These totals assume known, fixed outcomes and a short episode. A real agent generally does **not** know the consequences in advance: it must learn from interaction, and its actions can change later states. A model-based method can plan using a learned transition model; a model-free method can learn a policy or values directly from experience. Neither method makes an unsafe reward design safe.
