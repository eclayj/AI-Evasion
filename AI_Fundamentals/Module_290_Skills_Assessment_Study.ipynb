## 1. Bayes' theorem + classification → Naive Bayes

Read P(spam|flagged-word) as **probability of spam given the flagged word**. Bayes' theorem combines a prior with evidence. The 'naive' extension treats features as conditionally independent given a class; that assumption is a simplification, not a guarantee about real language. Here one feature is enough to expose the posterior arithmetic.

```python
prior_spam = 0.2
word_given_spam = 0.8
word_given_not_spam = 0.1
evidence = word_given_spam * prior_spam + word_given_not_spam * (1 - prior_spam)
posterior_spam = word_given_spam * prior_spam / evidence
print('P(spam | word) =', posterior_spam)
assert abs(posterior_spam - (2 / 3)) < 1e-12
```

## 2. Variance-preserving dimension reduction → PCA

Read PCA as **principal component analysis**. It searches for directions of high variance, then projects data onto selected directions. This toy data varies only in its first coordinate, so the second coordinate carries no variance. High variance is not automatically useful for classification.

```python
points = [(0.0, 5.0), (1.0, 5.0), (2.0, 5.0)]
mean_x = sum(point[0] for point in points) / len(points)
mean_y = sum(point[1] for point in points) / len(points)
variance_x = sum((point[0] - mean_x) ** 2 for point in points) / len(points)
variance_y = sum((point[1] - mean_y) ** 2 for point in points) / len(points)
print('variance by coordinate:', variance_x, variance_y)
assert variance_x > 0 and variance_y == 0
```

## 3. Greedy next-state value → Q-learning

Read Q(s,a) as **Q of state s and action a**. Q-learning's target uses max over next actions; SARSA instead uses the next action that the current policy actually selected. Read α as **alpha, the learning rate** and γ as **gamma, the discount factor**.

```python
old_q, reward, gamma, alpha = 0.2, 0.5, 0.9, 0.1
next_action_values = (1.0, -0.3)
target = reward + gamma * max(next_action_values)
new_q = old_q + alpha * (target - old_q)
print('Q-learning target:', target, 'updated Q:', new_q)
assert abs(new_q - 0.32) < 1e-12
```

## 4. Weights + bias + activation → neuron

Read z=Σᵢwᵢxᵢ+b as **z equals the sum of w sub i times x sub i plus bias**. The activation acts on z. A step activation is the classical perceptron example; ReLU is another activation, not another weighted sum.

```python
inputs, weights, bias = (2.0, -1.0), (0.5, 1.0), 0.0
z = sum(weight * value for weight, value in zip(weights, inputs)) + bias
relu_output = max(0.0, z)
print('pre-activation:', z, 'ReLU output:', relu_output)
assert z == 0 and relu_output == 0
```

## 5. Self-attention across a sequence → transformer

A transformer uses attention to weight information from sequence positions. An RNN instead carries a hidden state from one step to the next. Read softmax as **normalize scores into nonnegative weights that sum to one**. The two untrained scores below illustrate normalization, not language understanding.

```python
import math
scores = (1 / math.sqrt(2), 0.0)
unnormalized = [math.exp(score) for score in scores]
attention_weights = [value / sum(unnormalized) for value in unnormalized]
print('two-position attention weights:', attention_weights)
assert abs(sum(attention_weights) - 1.0) < 1e-12
assert attention_weights[0] > attention_weights[1]
```
