# Generative-AI concepts: sampling and mode collapse

Companion to HTB Academy Module 290, Section 21. This is an intentionally tiny **categorical distribution**, not a trained generative architecture. It isolates what it means to sample from a distribution and what goes wrong when a generator emits only one kind of output.

Read P(X=A)=0.5 as **the probability that random variable X equals A is one half**. Probabilities across all possible outcomes must total one. A sample is one random draw; proportions become meaningful over many draws, not for an individual draw.

```python
from collections import Counter
from random import Random

outcomes = ('A', 'B', 'C')
reference_weights = (0.5, 0.3, 0.2)  # An invented three-outcome distribution.
collapsed_weights = (1.0, 0.0, 0.0)  # Only A: no variety remains.
assert abs(sum(reference_weights) - 1.0) < 1e-12
assert abs(sum(collapsed_weights) - 1.0) < 1e-12

def draw_many(weights, seed, count=1000):
    """Return seeded independent draws, with no training step."""
    rng = Random(seed)
    return rng.choices(outcomes, weights=weights, k=count)

reference_draws = draw_many(reference_weights, seed=7)
collapsed_draws = draw_many(collapsed_weights, seed=7)
reference_counts = Counter(reference_draws)
collapsed_counts = Counter(collapsed_draws)
print('reference-like sample counts:', reference_counts)
print('collapsed sample counts:', collapsed_counts)
assert set(reference_draws) == set(outcomes)
assert set(collapsed_draws) == {'A'}
```

## Compare diversity and frequency

For this invented distribution, the reference-like sampler should produce A, B, and C; the collapsed one produces only A. A sample frequency is an estimate: count(A)/number of draws, read **count of A divided by the number of draws**. The test uses a broad tolerance because random samples vary even with a fixed probability distribution. We are not computing HTB's image or text quality metrics; those require actual model outputs and appropriate reference data.

```python
sample_count = len(reference_draws)
for outcome, expected_probability in zip(outcomes, reference_weights):
    observed_fraction = reference_counts[outcome] / sample_count
    print(f'{outcome}: expected {expected_probability:.1f}, observed {observed_fraction:.3f}')
    assert abs(observed_fraction - expected_probability) < 0.07
print('distinct outcomes:', len(reference_counts), 'versus', len(collapsed_counts))
assert len(reference_counts) == 3 and len(collapsed_counts) == 1
```
