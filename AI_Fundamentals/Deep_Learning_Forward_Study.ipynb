# Deep-learning primer: one forward calculation

Companion to HTB Academy Module 290, Section 16. This notebook makes the page's weighted connections and named activation functions visible with invented numbers. It is a **forward-pass illustration only**: no trained model, loss, gradient, or optimizer appears here. The broader neural-network training details belong to later lessons.

Read z = Σᵢ wᵢxᵢ + b as **z equals the sum of weight sub i times input sub i, plus bias b**. The sum is over input features. Read a = f(z) as **a equals activation f applied to z**.

```python
import math

# One invented input row: two features, one weight for each feature.
inputs = [2.0, -1.0]
weights = [0.5, 1.0]
bias = 0.0
assert len(inputs) == len(weights)
# z = sum_i (w_i * x_i) + b. zip pairs corresponding feature and weight.
weighted_terms = [weight * value for weight, value in zip(weights, inputs)]
z = sum(weighted_terms) + bias
print('weighted terms:', weighted_terms, 'pre-activation z:', z)
assert weighted_terms == [1.0, -1.0]
assert z == 0.0
```

## Compare the three activations named by HTB

ReLU returns max(0,z). Sigmoid smoothly squashes finite inputs between 0 and 1. Tanh smoothly squashes them between −1 and 1. These outputs are not interchangeable probabilities: a sigmoid output may be interpreted probabilistically only with the appropriate model, training objective, and calibration. The five inputs below are just test numbers, not data from a security task.

```python
def relu(value):
    return max(0.0, value)  # HTB's piecewise ReLU definition.

def sigmoid(value):
    return 1.0 / (1.0 + math.exp(-value))  # Standard sigmoid illustration.

for value in (-3.0, -1.0, 0.0, 1.0, 3.0):
    print(f'z={value:+.0f} | ReLU={relu(value):.3f} sigmoid={sigmoid(value):.3f} tanh={math.tanh(value):.3f}')
assert relu(-3) == 0 and relu(3) == 3
assert sigmoid(0) == 0.5 and math.tanh(0) == 0
assert all(0 < sigmoid(v) < 1 for v in (-3, -1, 0, 1, 3))
```

The computed z for our one neuron is zero, so its ReLU output would be 0, its sigmoid output 0.5, and its tanh output 0. The next layer, if there were one, would consume these activation outputs, not the original features. None of these toy outputs proves a network has learned useful or robust features.
