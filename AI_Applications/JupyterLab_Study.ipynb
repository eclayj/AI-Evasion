## A synthetic table and scatter plot

HTB uses independent random columns to demonstrate pandas and Matplotlib. For a sampled number u between 0 and 1, column2 = 10u, read **column two equals ten times u**. Thus u = 0.37 becomes 3.7. There is no predicted target and no model fitting here. The fixed seed makes the same 50-row example repeatable, which is a teaching addition to HTB's snippet.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(292)  # Teaching addition: reproducible random samples.
data = pd.DataFrame({
    "column1": rng.random(50),       # 50 independent values in [0, 1).
    "column2": rng.random(50) * 10,  # Scale another 50 values into [0, 10).
})
print(data.head())  # First five rows of this labeled table.

plt.scatter(data["column1"], data["column2"])
plt.xlabel("Column 1")
plt.ylabel("Column 2")
plt.title("Synthetic Scatter Plot (not a model)")
plt.show()  # Display the figure in Jupyter; no image file is written.
```
